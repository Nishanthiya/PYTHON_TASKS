```python
#task1 
my_list = [1, 2, 'example', [1, 2]]

result = []

for item in my_list:
    if isinstance(item, list):
        result.extend(item)
    else:
        result.append(item)

print(result)
```

```python
#task2 Find 2nd largest number

scores = [85, 92, 78, 92, 88]
largest = second = float('-inf')
for num in scores:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num

print(second)
```

```python
# Armstrong Number

num = int(input("Enter num:"))
STRING = str(num)
root = len(STRING)
amst = 0
for i in str(num):
    amst += int(i) ** root
if amst == num:
    print("It is Armstrong")
else:
    print("It is not Armstrong")

```
