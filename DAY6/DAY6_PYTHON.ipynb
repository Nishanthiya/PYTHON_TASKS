```python

# Task 1: Dictionary Lookup with Default Values
stock = {'LED': 20, 'RESISTOR': 50, 'SENSOR': 4}
requested = ['LED', 'SENSOR', 'SWITCH']
for item in requested:
    quantity = stock.get(item)

    if quantity is not None:
        print(item, ":", quantity, "available")
    else:
        print(item, ": unavailable")

```

```python
# Task 2: Set Operations on Course Registrations
A = {101, 102, 103, 105}
B = {102, 104, 105, 106}
print("Both courses:", sorted(A & B))
print("Only Course A:", sorted(A - B))
print("Either course:", sorted(A | B))

```

```python
# Task 3: Invert a Dictionary

marks = {'Arun': 85, 'Priya': 92, 'Kavin': 78}

result = {}

for name, mark in marks.items():
    result[mark] = name

print(result)

```

```python
# Task 4: Group Names by First Letter
names = ['Anu', 'Arun', 'Bala', 'Banu', 'Cathy', 'Asha']
result = {}

for name in names:
    letter = name[0].upper()

    if letter not in result:
        result[letter] = []

    result[letter].append(name)

print(result)

```

```python
# Task 5: Merge Inventory Dictionaries

store_a = {'LED': 20, 'Cable': 10, 'Board': 3}
store_b = {'Cable': 5, 'Sensor': 7, 'LED': 8}

combined = store_a.copy()
for item, quantity in store_b.items():
    if item in combined:
        combined[item] += quantity
    else:
        combined[item] = quantity
print("Combined inventory:", combined)
```

```python
# Task 6: Rank Products by Total Sales
sales = {
    'Sensor': [4, 6, 5],
    'Cable': [10, 2, 3],
    'Board': [8, 7],
    'LED': [5, 5, 5]}
total = {}
for product, quantities in sales.items():
    total[product] = sum(quantities)
for product, quantity in sorted(total.items(), key=lambda x: (-x[1], x[0])):
    print(product, ":", quantity)

```

```python
# Task 7: Build a Nested Dictionary from Device Records
records = [
    ('D01', 'Lab-A', 'READY'),
    ('D02', 'Lab-B', 'BUSY'),
    ('D01', 'Lab-C', 'MAINTENANCE')]
devices = {}
for device_id, location, status in records:
    devices[device_id] = {
        'location': location,
        'status': status }
print(devices)
```

```python
# Task 8: Find Common Permissions Across Users
permissions = {
    'admin': {'read', 'write', 'delete'},
    'operator': {'read', 'write'},
    'auditor': {'read', 'export'}}
common = set.intersection(*permissions.values())
all_permissions = set.union(*permissions.values())
print("Common permissions:", sorted(common))
print("All permissions:", sorted(all_permissions))
```

```python
# Task 9: Subarray Sum Equals K
nums = [1, 1, 1]
k = 2
count = 0
prefix_sum = 0
freq = {0: 1}
for num in nums:
    prefix_sum += num
    count += freq.get(prefix_sum - k, 0)
    freq[prefix_sum] = freq.get(prefix_sum, 0) + 1
print("Count:", count)
```

```python

# Task 10: Minimum Window Substring

s = 'ADOBECODEBANC'
t = 'ABC'

need = {}

for ch in t:
    need[ch] = need.get(ch, 0) + 1

left = 0
count = 0
start = 0
min_len = float('inf')
window = {}

for right in range(len(s)):
    ch = s[right]
    window[ch] = window.get(ch, 0) + 1

    if ch in need and window[ch] <= need[ch]:
        count += 1

    while count == len(t):
        if right - left + 1 < min_len:
            min_len = right - left + 1
            start = left

        left_ch = s[left]
        window[left_ch] -= 1

        if left_ch in need and window[left_ch] < need[left_ch]:
            count -= 1

        left += 1

if min_len == float('inf'):
    print("Minimum window:", '')
else:
    print("Minimum window:", s[start:start + min_len])

```
