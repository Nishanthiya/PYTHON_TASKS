```python
def analyze_students(records):
    qualified = []
    topper = ""
    highest_total = 0

    for student in records:
        name = student[0]
        total = sum(student[1:])
        average = total / 3

        if average >= 75:
            qualified.append(name)

        if total > highest_total:
            highest_total = total
            topper = name

    return {
        "qualified": qualified,
        "topper": topper
    }


records = [
    ("Asha", 85, 78, 92),
    ("Bala", 65, 72, 70),
    ("Charan", 90, 88, 95),
    ("Divya", 76, 80, 74),
    ("Esha", 60, 68, 72)
]

print(analyze_students(records))
```

```python
# Task 2: Remove Duplicates While Preserving Order

numbers = [10, 20, 10, 30, 20, 40, 30, 50]

unique_numbers = []

for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)

print(unique_numbers)
```

```python
# Task 3: Word Frequency Counter

text = "Python is easy and Python is powerful"

words = text.lower().split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
# Task 4: Filter and Transform Data

marks = [45, 82, 67, 91, 76, 55, 88]

result = [x**2 for x in marks if x >= 70]

print(result)
```

```python
# Task 5: Two Sum

nums = [2, 7, 11, 15]
target = 9

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            print([i, j])
```

```python
# Task 9: Group Anagrams

words = ["eat", "tea", "tan", "ate", "nat", "bat"]

groups = {}

for word in words:
    key = "".join(sorted(word))

    if key not in groups:
        groups[key] = []

    groups[key].append(word)

print(list(groups.values()))
```

```python
# Task 10: Customer Data Cleaning

data = [
    " Asha, Chennai, 25 ",
    "Bala, chennai, 30",
    " asha, Chennai, 25",
    "Charan, Bangalore, 28",
    "Bala, Chennai, 30 "
]

cleaned = []

for record in data:
    name, city, age = record.strip().split(",")

    name = name.strip().title()
    city = city.strip().title()
    age = int(age.strip())

    student = {
        "name": name,
        "city": city,
        "age": age
    }

    if student not in cleaned:
        cleaned.append(student)

print(cleaned)
```
