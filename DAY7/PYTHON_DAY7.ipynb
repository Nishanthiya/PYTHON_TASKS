```python
def product_except_self(nums):
    result = []

    for i in range(len(nums)):
        product = 1

        for j in range(len(nums)):
            if i != j:
                product *= nums[j]

        result.append(product)

    return result


nums = [1, 2, 3, 4]

print("Product array:", product_except_self(nums))
```

```python
def calculate_bill(prices, tax_rate=0.10, discount=0.20):
    subtotal = sum(prices)

    discount_amount = subtotal * discount
    discounted_price = subtotal - discount_amount

    tax = discounted_price * tax_rate
    total = discounted_price + tax

    print("Subtotal:", f"{subtotal:.2f}")
    print("Discount:", f"{discount_amount:.2f}")
    print("Tax:", f"{tax:.2f}")
    print("Final total:", f"{total:.2f}")


prices = [100, 50]

calculate_bill(prices, tax_rate=0.10, discount=0.20)
```

```python
def average_readings(*args):
    if len(args) == 0:
        return None

    average = sum(args) / len(args)
    return average


print("Average:", average_readings(10, 20, 30, 40))

if average_readings() is None:
    print("No readings provided")
```

```python
def create_device_config(device_id, **kwargs):
    config = {"device_id": device_id}

    allowed = ["baud_rate", "parity", "timeout_ms", "debug"]

    for key, value in kwargs.items():
        if key not in allowed:
            raise ValueError("Unsupported key: " + key)

        config[key] = value

    return config


result = create_device_config(
    "D01",
    baud_rate=115200,
    parity="N",
    timeout_ms=500
)

print(result)
```

```python
def group_anagrams(strs):
    groups = {}

    for word in strs:
        key = "".join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())


strs = ["eat", "tea", "tan", "ate", "nat", "bat"]

print("Groups:")
for group in group_anagrams(strs):
    print(group)
```

```python
def top_k_frequent(nums, k):
    freq = {}

    for num in nums:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1

    sorted_nums = sorted(freq, key=freq.get, reverse=True)

    return sorted_nums[:k]


nums = [1, 1, 1, 2, 2, 3]
k = 2

print("Top", k, "frequent values:", top_k_frequent(nums, k))
```

```python
def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        if num - 1 not in num_set:
            count = 1

            while num + count in num_set:
                count += 1

            longest = max(longest, count)

    return longest


nums = [100, 4, 200, 1, 3, 2]

print("Longest consecutive sequence length:", longest_consecutive(nums))
print("Sequence: 1, 2, 3, 4")
```

```python
def find_max_length(nums):
    balance = 0
    seen = {0: -1}
    max_length = 0

    for i in range(len(nums)):
        if nums[i] == 1:
            balance += 1
        else:
            balance -= 1

        if balance in seen:
            length = i - seen[balance]
            max_length = max(max_length, length)
        else:
            seen[balance] = i

    return max_length


nums = [0, 1, 0]

print("Maximum balanced subarray length:", find_max_length(nums))
```

```python
def length_of_last_word(s):
    words = s.strip().split()

    if len(words) == 0:
        return 0

    return len(words[-1])


s = " Embedded Python "

print("Length of last word:", length_of_last_word(s))
```
