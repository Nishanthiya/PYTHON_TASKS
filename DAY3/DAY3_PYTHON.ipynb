```python

# Task Q1: Type Casting & Numeric Operations

try:
    num1 = input("Enter Integer: ")
    num2 = input("Enter Decimal: ")

    integer = int(num1)
    decimal = float(num2)

    print("Integer Value:", integer)
    print("Integer Type:", type(integer).__name__)

    print("Float Value:", decimal)
    print("Float Type:", type(decimal).__name__)

    print("Rounded Value:", round(decimal))

    if integer != 0:
        print(integer, "/", int(decimal), "=", round(integer / int(decimal), 4))
        print(integer, "//", int(decimal), "=", integer // int(decimal))
        print(integer, "%", int(decimal), "=", integer % int(decimal))

    print(integer, "** 2 =", integer ** 2)

except ValueError:
    print("Invalid numeric input!")

```

```python

# Task Q2: Caesar Cipher

message = input("Enter Message: ")
shift = int(input("Enter Shift: "))

encrypted = ""

for ch in message:
    if 'A' <= ch <= 'Z':
        new_char = chr((ord(ch) - ord('A') + shift) % 26 + ord('A'))
        encrypted += new_char

    elif 'a' <= ch <= 'z':
        new_char = chr((ord(ch) - ord('a') + shift) % 26 + ord('a'))
        encrypted += new_char

    else:
        encrypted += ch

print("Encrypted:", encrypted)

decrypted = ""

for ch in encrypted:
    if 'A' <= ch <= 'Z':
        new_char = chr((ord(ch) - ord('A') - shift) % 26 + ord('A'))
        decrypted += new_char

    elif 'a' <= ch <= 'z':
        new_char = chr((ord(ch) - ord('a') - shift) % 26 + ord('a'))
        decrypted += new_char

    else:
        decrypted += ch

print("Decrypted:", decrypted)

```

```python

# Task Q3: Number Analyzer

start = int(input("Start: "))
end = int(input("End: "))

print("Number Prime Perfect Armstrong Palindrome DigitSum Digits Binary")

for n in range(start, end + 1):
    # Prime check
    prime = n > 1
    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    # Perfect number check
    total = 0
    for i in range(1, n):
        if i > 0 and n % i == 0:
            total += i
    perfect = n > 1 and total == n

    # Digit operations
    digits = str(n)
    digit_sum = sum(int(d) for d in digits)
    digit_count = len(digits)

    # Armstrong number check
    power = len(digits)
    armstrong_sum = sum(int(d) ** power for d in digits)
    armstrong = armstrong_sum == n

    # Palindrome check
    palindrome = digits == digits[::-1]

    # Binary representation using loops
    temp = n
    binary = ""

    if temp == 0:
        binary = "0"
    else:
        while temp > 0:
            binary = str(temp % 2) + binary
            temp //= 2

    print(
        n,
        "Yes" if prime else "No",
        "Yes" if perfect else "No",
        "Yes" if armstrong else "No",
        "Yes" if palindrome else "No",
        digit_sum,
        digit_count,
        binary
    )

```

```python

# Task Q4: Number System Converter + GCD/LCM

num1 = int(input("Number 1: "))
num2 = int(input("Number 2: "))

# Decimal to Binary
n = num1
binary = ""

while n > 0:
    binary = str(n % 2) + binary
    n //= 2

if binary == "":
    binary = "0"

# Decimal to Octal
n = num1
octal = ""

while n > 0:
    octal = str(n % 8) + octal
    n //= 8

if octal == "":
    octal = "0"

# Decimal to Hexadecimal
digits = "0123456789ABCDEF"
n = num1
hexa = ""

while n > 0:
    remainder = n % 16
    hexa = digits[remainder] + hexa
    n //= 16

if hexa == "":
    hexa = "0"

# GCD using Euclidean algorithm
a = num1
b = num2

while b != 0:
    a, b = b, a % b

gcd = a

# LCM
lcm = (num1 * num2) // gcd if gcd != 0 else 0

print("Binary:", binary)
print("Octal:", octal)
print("Hexadecimal:", hexa)
print("GCD:", gcd)
print("LCM:", lcm)

```

```python

# Task Q5: Two-Divisor FizzBuzz

n = int(input("N: "))

d1 = int(input("Divisor 1: "))
w1 = input("Word 1: ")

d2 = int(input("Divisor 2: "))
w2 = input("Word 2: ")

for i in range(1, n + 1):

    if i % d1 == 0 and i % d2 == 0:
        print(w1 + w2)

    elif i % d1 == 0:
        print(w1)

    elif i % d2 == 0:
        print(w2)

    else:
        print(i)

```

```python

# Task Q6: Student Mark Conversion & Result Analyzer

students = int(input("Students: "))

for i in range(students):
    name = input("Enter student name: ")
    marks_input = input("Enter marks separated by spaces: ")

    try:
        marks = [float(x) for x in marks_input.split()]

        if len(marks) == 0 or any(m < 0 or m > 100 for m in marks):
            print(name, "Invalid marks")
            continue

        total = sum(marks)
        average = total / len(marks)

        if average >= 90:
            grade = "A"
        elif average >= 75:
            grade = "B"
        elif average >= 60:
            grade = "C"
        elif average >= 50:
            grade = "D"
        else:
            grade = "F"

        print("\nStudent:", name)
        print("Total:", total)
        print("Average:", format(average, ".2f"))
        print("Grade:", grade)

    except ValueError:
        print(name, "Invalid marks")

```

```python

# Task Q7: Digital Payment Transaction Analyzer

n = int(input("Transactions: "))

valid = []
rejected = 0

for i in range(n):
    amount = input("Enter transaction amount: ")

    try:
        amount = float(amount)

        if amount < 0:
            rejected += 1
        else:
            valid.append(amount)

    except ValueError:
        rejected += 1

if len(valid) > 0:
    total = sum(valid)
    highest = max(valid)
    lowest = min(valid)
    average = total / len(valid)

    print("Successful Transactions:", len(valid))
    print("Total Amount: ₹{:.2f}".format(total))
    print("Highest Transaction: ₹{:.2f}".format(highest))
    print("Lowest Transaction: ₹{:.2f}".format(lowest))
    print("Average Transaction: ₹{:.2f}".format(average))

else:
    print("No successful transactions.")

print("Rejected Transactions:", rejected)

```

```python

# Task Q8: LeetCode #8 - String to Integer (atoi)

s = input("Enter a string: ")

i = 0
n = len(s)
sign = 1
result = 0

# Skip leading spaces
while i < n and s[i] == ' ':
    i += 1

# Check the sign
if i < n and (s[i] == '+' or s[i] == '-'):
    if s[i] == '-':
        sign = -1
    i += 1

# Convert digits
while i < n and s[i].isdigit():
    digit = ord(s[i]) - ord('0')
    result = result * 10 + digit
    i += 1

print(sign * result)

```

```python

# Task Q9: LeetCode #67 - Add Binary

a = input("Enter first binary number: ")
b = input("Enter second binary number: ")

i = len(a) - 1
j = len(b) - 1
carry = 0
result = ""

while i >= 0 or j >= 0 or carry:
    bit1 = int(a[i]) if i >= 0 else 0
    bit2 = int(b[j]) if j >= 0 else 0

    total = bit1 + bit2 + carry

    result = str(total % 2) + result
    carry = total // 2

    i -= 1
    j -= 1

print("Sum:", result)

```

```python

# Task Q10: LeetCode #202 - Happy Number

n = int(input("Enter a number: "))

seen = set()

while n != 1 and n not in seen:
    seen.add(n)

    total = 0

    while n > 0:
        digit = n % 10
        total += digit ** 2
        n //= 10

    n = total

if n == 1:
    print("True")
else:
    print("False")

```
