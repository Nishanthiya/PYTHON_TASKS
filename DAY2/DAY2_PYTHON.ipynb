```python
# Task 1: Prime, Factorial, Fibonacci, Digit Sum & Reverse

n = int(input("Enter N: "))
number = int(input("Enter a number: "))

# Print prime numbers up to N
print("Primes:", end=" ")
for i in range(2, n + 1):
    prime = True
    for j in range(2, i):
        if i % j == 0:
            prime = False
            break
    if prime:
        print(i, end=" ")

# Calculate factorial
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print("\nFactorial:", factorial)

# Print Fibonacci numbers
a, b = 0, 1
print("Fibonacci:", end=" ")
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b

# Calculate digit sum and reverse
temp = abs(number)
digit_sum = 0
reverse = 0

while temp > 0:
    digit = temp % 10
    digit_sum += digit
    reverse = reverse * 10 + digit
    temp //= 10

print("\nDigit Sum:", digit_sum)
if number < 0:
    reverse = -reverse
print("Reverse:", reverse)
```

```python
# Task 2: Patterns & Multiplication Grid

n = int(input("Enter N: "))

# Right Triangle
print("\nRight Triangle")
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()

# Pyramid
print("\nPyramid")
for i in range(1, n + 1):
    for j in range(n - i):
        print(" ", end="")
    for j in range(2 * i - 1):
        print("*", end="")
    print()

# Number Triangle
print("\nNumber Triangle")
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()

# Multiplication Grid (1 to 10)
print("\nMultiplication Grid")
for i in range(1, 11):
    for j in range(1, 11):
        print(f"{i} x {j} = {i*j}", end="\t")
    print()

```

```python
# Task 3: String Analyzer

text = input("Enter a string: ")

# Reverse using a loop
reverse = ""
for ch in text:
    reverse = ch + reverse

print("Loop Reverse:", reverse)

# Reverse using slicing
print("Slice Reverse:", text[::-1])

# Check palindrome (ignore case and spaces)
cleaned = text.replace(" ", "").lower()

if cleaned == cleaned[::-1]:
    print("Palindrome: Yes")
else:
    print("Palindrome: No")

# Count vowels, consonants and digits
vowels = 0
consonants = 0
digits = 0

for ch in text.lower():
    if ch in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
```

```python
# Task 4: Formatted Electricity Bill

customer = input("Enter customer name: ")
units = int(input("Enter units consumed: "))

# Calculate electricity bill using slabs
if units <= 100:
    amount = units * 3
elif units <= 200:
    amount = 100 * 3 + (units - 100) * 5
elif units <= 300:
    amount = 100 * 3 + 100 * 5 + (units - 200) * 7
else:
    amount = 100 * 3 + 100 * 5 + 100 * 7 + (units - 300) * 10

# Print formatted bill
print("\n----- Electricity Bill -----")
print(f"Customer : {customer}")
print(f"Units    : {units}")
print(f"Amount   : Rs.{amount:.2f}")
```

```python
#5 ATM Cash Withdrawal & Denomination Breakdown

amount = int(input("Enter withdrawal amount: "))

if amount <= 0 or amount > 20000 or amount % 10 != 0:
    print("Invalid withdrawal amount")
else:
    remaining = amount
    total_notes = 0

    for denomination in [500, 200, 100, 50, 10]:
        notes = remaining // denomination
        remaining %= denomination
        total_notes += notes
        print(f"Rs.{denomination} notes : {notes}")

    print("Total Notes :", total_notes)
    print(f"Amount      : Rs.{amount}")
```

```python
#6 Digital Parking Fee & Time Calculator

entry = input("Enter entry time (HH:MM): ")
exit_time = input("Enter exit time (HH:MM): ")

eh, em = map(int, entry.split(":"))
xh, xm = map(int, exit_time.split(":"))

start = eh * 60 + em
end = xh * 60 + xm

if end < start:
    end += 24 * 60

duration = end - start
hours = duration // 60
minutes = duration % 60

billable_hours = (duration + 59) // 60

if duration == 0:
    fee = 0
else:
    fee = 30 + max(0, billable_hours - 1) * 20

print(f"Parking Duration : {hours} hours {minutes} minutes")
print("Billable Hours   :", billable_hours)
print(f"Parking Fee      : Rs.{fee}")
```

```python
#7 Delivery Charge & Earnings Report


n = int(input("Number of deliveries: "))

total_distance = 0
total_earnings = 0

for i in range(1, n + 1):
    distance = float(input(f"Enter distance for delivery {i}: "))
    total_distance += distance

    if distance <= 5:
        earnings = 40
    else:
        earnings = 40 + (distance - 5) * 8

    total_earnings += earnings

if n > 0:
    average = total_distance / n
else:
    average = 0

print(f"Total Distance   : {total_distance:.2f} km")
print(f"Total Earnings   : Rs.{total_earnings:.2f}")
print(f"Average Distance : {average:.2f} km")
```

```python
#8 — LeetCode #7: Reverse Integer


x = int(input("Enter an integer: "))

sign = -1 if x < 0 else 1
x = abs(x)
reverse = 0

while x > 0:
    digit = x % 10
    reverse = reverse * 10 + digit
    x //= 10

print(sign * reverse)
```

```python
#9 — LeetCode #9: Palindrome Number


x = int(input("Enter an integer: "))

if x < 0:
    print(False)
else:
    original = x
    reverse = 0

    while x > 0:
        digit = x % 10
        reverse = reverse * 10 + digit
        x //= 10

    print(original == reverse)
```

```python
#10 — LeetCode #412: Fizz Buzz

n = int(input("Enter N: "))

for i in range(1, n + 1):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```
